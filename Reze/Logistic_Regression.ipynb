# Logistic Regression — Simple and Intuitive Mathematical Explanation


## 1. Intuition

We want to **classify** data into two groups:

- $y = 1$ (positive class)
- $y = 0$ (negative class)

A linear function $z = w^\top x + b$ can be any number from $-\infty$ to $+\infty$,
but we want a **probability** between 0 and 1.

So we pass the linear output through a *sigmoid* function, which squashes it into a
probability in $[0, 1]$.


## 2. The Sigmoid Function

The sigmoid turns any number into a probability:

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

- If $z$ is big and positive → $\sigma(z) \approx 1$
- If $z$ is big and negative → $\sigma(z) \approx 0$
- If $z = 0$ → $\sigma(z) = 0.5$

It is perfect for classification.


## 3. The Logistic Regression Model

We first compute:

$$
z = w^\top x + b
$$

Then convert it to a probability:

$$
h(x) = \sigma(z)
$$

This means:

- If $h(x)$ is close to 1 → predict class 1  
- If $h(x)$ is close to 0 → predict class 0


## 4. Loss Function (How the Model Learns)

We want the predicted probability $h(x)$ to be close to the true label $y$.

If the true label is:
- $y = 1$, we want $h(x)$ to be large.
- $y = 0$, we want $h(x)$ to be small.

The loss for one example is:

$$
\text{loss} = -\left[
y \log(h(x)) + (1-y) \log(1 - h(x))
\right]
$$

This simply punishes the model when:
- it predicts a small number for a true label of 1,
- or predicts a large number for a true label of 0.


# 🔢 Logistic Loss Numerical Examples

The logistic (binary cross-entropy) loss is:

$$
\text{loss} = -\left[y \log(h(x)) + (1-y) \log(1 - h(x))\right]
$$

Where $y$ is the true label (0 or 1) and $h(x)$ is the predicted probability.

| True Label $y$ | Predicted Probability $h(x)$ | Loss | Interpretation |
|----------------|-----------------------------|------|----------------|
| 1 | 0.95 | 0.051 | Correct & confident → very low loss |
| 1 | 0.2  | 1.609 | Wrong prediction → high loss |
| 0 | 0.8  | 1.609 | Wrong prediction → high loss |
| 0 | 0.05 | 0.051 | Correct & confident → very low loss |

**Key Idea:**  
- Loss is small when the predicted probability is close to the true label  
- Loss is large when the predicted probability is far from the true label


## 5. Total Cost (for all training examples)

If we have $n$ examples:

$$
J(w,b) = -\sum_{i=1}^n
\left[
y^{(i)} \log h(x^{(i)}) 
+ 
(1-y^{(i)}) \log (1 - h(x^{(i)}))
\right]
$$

This is the function we want to **minimize**.


## 6. Gradient (How Much We Should Change the Parameters)

For each training example, the model computes:

$$
\text{error} = h(x) - y
$$

This error tells us if:
- the prediction is too high,
- or too low.

The gradients are:

$$
\frac{\partial J}{\partial w} = \text{error} \cdot x
$$

$$
\frac{\partial J}{\partial b} = \text{error}
$$

Very intuitive:
- If prediction is wrong → update more.
- If prediction is right → update less.


## 7. Gradient Descent (Learning Algorithm)

We update the parameters in the direction that reduces the cost:

$$
w \leftarrow w - \alpha \frac{\partial J}{\partial w}
$$

$$
b \leftarrow b - \alpha \frac{\partial J}{\partial b}
$$

Where $\alpha$ is the learning rate.

Repeat until the model learns the correct boundary.


## 8. Decision Boundary

We classify based on:

$$
h(x) \ge 0.5 \quad \Rightarrow \quad y=1
$$

Because:

$$
h(x) = 0.5 \quad \Leftrightarrow \quad w^\top x + b = 0
$$

So the **decision boundary** is:

- A **line** in 2D
- A **plane** in 3D
- A **hyperplane** in higher dimensions
